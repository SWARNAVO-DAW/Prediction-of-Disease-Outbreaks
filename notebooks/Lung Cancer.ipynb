## _Import Libraries_

```python
import numpy as np
import pandas as pd
import pickle

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
df = pd.read_csv('../data/survey lung cancer.csv')
```

# _Clean the dataset_

## _Clean column names_

```python
df.columns = df.columns.str.strip()
```

## _Check the column names_

```python
print("Column names:")
print(df.columns.tolist())
```

## _Check missing values_

```python
print("\nMissing values:")
print(df.isnull().sum())
```

## _Check duplicate rows_

```python
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

## Encode GENDER and LUNG_CANCER

## _Encode Gender_

```python
df['GENDER'] = df['GENDER'].map({
    'M': 1,
    'F': 0
})
```

## _Encode target variable_

```python
df['LUNG_CANCER'] = df['LUNG_CANCER'].map({
    'YES': 1,
    'NO': 0
})
```

## _Check the encoded values_

```python
print("GENDER values:")
print(df['GENDER'].value_counts())           #1 = YES , 0 = NO
```

```python
print("\nLUNG_CANCER values:")
print(df['LUNG_CANCER'].value_counts())
```

```python
# Check data types
print("Data types:")
print(df.dtypes)

# Check unique values in each column
print("\nUnique values:")
for column in df.columns:
    print(f"{column}: {df[column].unique()}")
```

## Splitting the Features and Target

## _Define input features_

```python
features = [
    'GENDER',
    'AGE',
    'SMOKING',
    'YELLOW_FINGERS',
    'ANXIETY',
    'PEER_PRESSURE',
    'CHRONIC DISEASE',
    'FATIGUE',
    'ALLERGY',
    'WHEEZING',
    'ALCOHOL CONSUMING',
    'COUGHING',
    'SHORTNESS OF BREATH',
    'SWALLOWING DIFFICULTY',
    'CHEST PAIN'
]
```

## _Create feature matrix_

```python
X = df[features].copy()
```

## _Create target variable_

```python
y = df['LUNG_CANCER'].copy()
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("\nFeatures:")
print(X.columns.tolist())
```

```python
print("\nTarget distribution:")
print(y.value_counts())
```

## Split the data into training and testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
print("\nTraining target distribution:")
print(y_train.value_counts())
```

```python
print("\nTesting target distribution:")
print(y_test.value_counts())
```

## Train the Lung Cancer Logistic Regression model

## _Create the Logistic Regression model_

```python
lung_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)
```

## _Train the model_

```python
lung_model.fit(X_train, y_train)

print("Lung Cancer Logistic Regression model trained successfully.")
```

## Evaluate the model

## _Make predictions on training and testing data_

```python
y_train_pred = lung_model.predict(X_train)
y_test_pred = lung_model.predict(X_test)
```

## _Calculate accuracy_

```python
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)
```

```python
print("Training Accuracy:", train_accuracy)
print("Testing Accuracy:", test_accuracy)
```

## Classification Report

## _Classification report for the test data_

```python
print("Classification Report:")
print(classification_report(
    y_test,
    y_test_pred,
    target_names=["NO", "YES"]
))
```

# Confusion Matrix

```python
cm = confusion_matrix(y_test, y_test_pred)

print("Confusion Matrix:")
print(cm)
```

# Try Random Forest

## _Train Random Forest model_

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight='balanced'
)

rf_model.fit(X_train, y_train)

# Predictions
rf_test_pred = rf_model.predict(X_test)

# Accuracy
rf_accuracy = accuracy_score(y_test, rf_test_pred)

print("Random Forest Testing Accuracy:", rf_accuracy)

print("\nRandom Forest Classification Report:")
print(classification_report(
    y_test,
    rf_test_pred,
    target_names=["NO", "YES"]
))
```

## _Random Forest Confusion Matrix_

```python
rf_cm = confusion_matrix(y_test, rf_test_pred)

print("Random Forest Confusion Matrix:")
print(rf_cm)
```

# Select Random Forest as the final model

## Final Model Selection

Based on the evaluation results, Random Forest performed better than Logistic Regression.

Random Forest Testing Accuracy: 90.32%

Therefore, Random Forest is selected as the final Lung Cancer prediction model.

## _Select Random Forest as the final model_

```python
final_model = rf_model

print("Random Forest selected as the final Lung Cancer model.")
```

## Save the Random Forest model

## _Save the final Lung Cancer model_

```python
with open('../models/lungs_disease_model.sav', 'wb') as file:
    pickle.dump(final_model, file)

print("Lung Cancer Random Forest model saved successfully.")
```

## _Reload the saved model_

```python
with open('../models/lungs_disease_model.sav', 'rb') as file:
    loaded_lung_model = pickle.load(file)

print("Saved Lung Cancer model loaded successfully.")
print("Model type:", type(loaded_lung_model))
```

## Test the reloaded model on the test dataset

## _Verify the reloaded model_

```python
loaded_test_pred = loaded_lung_model.predict(X_test)

loaded_accuracy = accuracy_score(y_test, loaded_test_pred)

print("Accuracy of reloaded model:", loaded_accuracy)
```

## Test one patient

```python
# Test prediction for a sample patient

new_patient = pd.DataFrame([{
    'GENDER': 1,
    'AGE': 60,
    'SMOKING': 2,
    'YELLOW_FINGERS': 2,
    'ANXIETY': 2,
    'PEER_PRESSURE': 1,
    'CHRONIC DISEASE': 2,
    'FATIGUE': 2,
    'ALLERGY': 1,
    'WHEEZING': 2,
    'ALCOHOL CONSUMING': 2,
    'COUGHING': 2,
    'SHORTNESS OF BREATH': 2,
    'SWALLOWING DIFFICULTY': 2,
    'CHEST PAIN': 2
}])

prediction = loaded_lung_model.predict(new_patient)

print("Prediction:", prediction[0])

if prediction[0] == 1:
    print("Result: YES - likely lung cancer")
else:
    print("Result: NO - lung cancer not predicted")
```
